Vector)การแปลงข้อมูลเชิงปริมาณของวัตถุหรือตัวตน (เช่น ลูกค้า 1 คน) ให้อยู่ในรูปเวกเตอร์ $d$-มิติ ใน ปริภูมิคุณลักษณะ (Feature Space):$$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_d \end{bmatrix} \in \mathbb{R}^d$$$x_i$ คือค่าของตัวแปรแต่ละตัว (เช่น อายุ, รายได้, ยอดซื้อ)ลำดับของมิติมีความสำคัญมาก การเปรียบเทียบหรือดำเนินกลยุทธ์ทางคณิตศาสตร์ระหว่างสองเวกเตอร์ต้องอ้างอิงลำดับตัวแปรชุดเดียวกันเสมอ2. ขนาดของเวกเตอร์ (L2 Norm / Vector Magnitude)ความยาวหรือระยะห่างของเวกเตอร์ $\mathbf{x}$ จากจุดกำเนิด (Origin: $\mathbf{0}$) คำนวณจาก L2 Norm (หรือ Euclidean Norm):$$\Vert{}\mathbf{x}\Vert{} = \sqrt{\sum_{i=1}^{d} x_i^2} = \sqrt{x_1^2 + x_2^2 + \dots + x_d^2}$$การประยุกต์ใช้: วัดขนาดรวมของพฤติกรรมลูกค้า เช่น ยอดรวมการใช้งานหรือยอดรวมการจับจ่ายในหลายๆ มิติ3. ผลคูณจุด (Dot Product) และแบบจำลองให้คะแนน (Linear Scoring Model)การนำสมาชิกในตำแหน่งเดียวกันของสองเวกเตอร์มาคูณกัน แล้วบวกผลคูณทั้งหมดเข้าด้วยกัน:$$\mathbf{a} \cdot \mathbf{b} = \mathbf{a}^T \mathbf{b} = \sum_{i=1}^{d} a_i b_i = a_1 b_1 + a_2 b_2 + \dots + a_d b_d$$ความสัมพันธ์เชิงเรขาคณิต: $\mathbf{a} \cdot \mathbf{b} = \Vert{}\mathbf{a}\Vert{} \Vert{}\mathbf{b}\Vert{} \cos \theta$การประยุกต์ใช้ในการถ่วงน้ำหนัก (Weighted Scoring): เมื่อคำนวณผลคูณจุดระหว่างเวกเตอร์ลักษณะ $\mathbf{x}$ กับเวกเตอร์น้ำหนักความสำคัญ $\mathbf{w} = [w_1, w_2, \dots, w_d]$ จะได้คะแนนรวมเชิงเส้น (Linear Combination Score):$$\text{Score} = \mathbf{x} \cdot \mathbf{w} = \sum_{i=1}^{d} w_i x_i$$4. ความคล้ายคลึงแบบโคไซน์ (Cosine Similarity) และมุมระหว่างเวกเตอร์การวัด ทิศทาง หรือ สัดส่วนพฤติกรรม ของเวกเตอร์สองตัวว่าไปในแนวทางเดียวกันมากน้อยเพียงใด โดยไม่คำนึงถึงขนาด (Scale/Magnitude) ของตัวเลข:$$\text{Cosine Similarity}(\mathbf{a}, \mathbf{b}) = \cos \theta = \frac{\mathbf{a} \cdot \mathbf{b}}{\Vert{}\mathbf{a}\Vert{} \Vert{}\mathbf{b}\Vert{}} = \frac{\sum_{i=1}^{d} a_i b_i}{\sqrt{\sum_{i=1}^{d} a_i^2} \sqrt{\sum_{i=1}^{d} b_i^2}}$$คุณสมบัติและการแปลงเป็นมุมองศาช่วงค่า: อยู่ในช่วง $[-1, 1]$ (สำหรับข้อมูลตัวเลขบวก ค่าจะอยู่ในช่วง $[0, 1]$)$\cos \theta = 1.0$ $\rightarrow$ เวกเตอร์ชี้ไปทิศทางเดียวกันสมบูรณ์ ($\theta = 0^\circ$)$\cos \theta = 0.0$ $\rightarrow$ เวกเตอร์ตั้งฉากกัน ไม่มีความเกี่ยวข้องกัน ($\theta = 90^\circ$)การคำนวณมุม ($\theta$):$$\theta = \arccos\left(\text{Cosine Similarity}\right) \times \frac{180^\circ}{\pi}$$5. ระยะทางแบบยุคลิด (Euclidean Distance)การวัดระยะห่างจริงในเชิงพื้นที่ระหว่างจุดสองจุดในปริภูมิ $d$-มิติ (วัดความต่างของตัวเลขจริง):$$d(\mathbf{a}, \mathbf{b}) = \Vert{}\mathbf{a} - \mathbf{b}\Vert{} = \sqrt{\sum_{i=1}^{d} (a_i - b_i)^2}$$ข้อแตกต่างระหว่าง Euclidean Distance และ Cosine SimilarityEuclidean Distance: ไวต่อขนาดตัวเลข (Magnitude Sensitive) เหมาะสำหรับหาจุดที่ตัวเลขใกล้เคียงกันจริงๆCosine Similarity: ไม่สนใจขนาดตัวเลข (Magnitude Invariant) เหมาะสำหรับหาอัตราส่วนพฤติกรรมที่คล้ายกัน6. เวกเตอร์ค่าเฉลี่ย (Mean Vector / Centroid)การหาจุดศูนย์กลางที่เป็นตัวแทนของกลุ่มข้อมูล (Cluster Center) $S$ คำนวณจากการหาค่าเฉลี่ยของข้อมูลในแต่ละมิติ:$$\boldsymbol{\mu} = \frac{1}{\vert{}S\vert{}} \sum_{i \in S} \mathbf{x}_i = \begin{bmatrix} \frac{1}{\vert{}S\vert{}} \sum x_{i,1} \\ \frac{1}{\vert{}S\vert{}} \sum x_{i,2} \\ \vdots \\ \frac{1}{\vert{}S\vert{}} \sum x_{i,d} \end{bmatrix}$$การประยุกต์ใช้: ใช้สร้าง โปรไฟล์กลุ่มลูกค้า (Customer Profile / Persona) เพื่อนำเวกเตอร์ค่าเฉลี่ยของแต่ละกลุ่มมาเปรียบเทียบกันด้วย Cosine Similarity หรือ Euclidean Distance7. เมทริกซ์ความคล้ายคลึง (Pairwise Similarity Matrix)เมื่อมีเวกเตอร์จำนวน $N$ ตัว การเปรียบเทียบความคล้ายคลึงระหว่างทุกคู่จะสร้างเป็นเมทริกซ์ขนาด $N \times N$:$$S_{ij} = \text{Cosine Similarity}(\mathbf{x}_i, \mathbf{x}_j)$$สมบัติสมมาตร (Symmetry): $S_{ij} = S_{ji}$แนวทแยงหลัก **ข้อ1.1
ให้ x = [tenure_months, web_visits, spend_k, orders_90d] เป็นเวกเตอร์ลักษณะของลูกค้า
แถวที่ 881 ความยาว (L2 norm) ของเวกเตอร์นี้ ‖x‖ =
(ทศนิยม 4 ตำแหน่ง)**

```python
import pandas as pd
import numpy as np

# 1. โหลดไฟล์ข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. เลือกลูกค้าแถวที่ 881 (นับจาก 0 ตามลำดับในไฟล์)
row_881 = df.iloc[881]

# 3. สร้าง Feature Vector x ตามลำดับที่โจทย์กำหนด: [tenure_months, web_visits, spend_k, orders_90d]
x = row_881[['tenure_months', 'web_visits', 'spend_k', 'orders_90d']].values.astype(np.float64)

# -------------------------------------------------------------
# วิธีที่ 1: คำนวณตรงๆ ตามสูตร sqrt( sum( x_i^2 ) )
# -------------------------------------------------------------
squared_elements = x ** 2          # [5^2, 7^2, 2.68^2, 1^2] -> [25.0, 49.0, 7.1824, 1.0]
sum_squared = np.sum(squared_elements) # ผลรวม = 82.1824
l2_norm_manual = np.sqrt(sum_squared)  # ถอด sqrt = 9.0654509...

# -------------------------------------------------------------
# วิธีที่ 2: ใช้ฟังก์ชันสำเร็จรูปของ NumPy (แนะนำ)
# -------------------------------------------------------------
l2_norm_numpy = np.linalg.norm(x)

# 4. แสดงผลลัพธ์แบบปัดทศนิยม 4 ตำแหน่ง
print(f"Feature Vector x: {x}")
print(f"คำตอบ L2 Norm: {l2_norm_numpy:.4f}")
```

**1.2 บริษัทประกันออกแพ็กเกจ "ครอบครัว" ที่รวมพฤติกรรมของลูกค้าสองคนเป็นบัญชีเดียว โดยนำเวกเตอร์ลักษณะ (รายได้ต่อเดือน, จำนวนอีเมลที่เปิดอ่าน, ขนาดตะกร้า, ยอดซื้อต่อเดือน) เรียงตามลำดับนี้ ของลูกค้าแถวที่ 898 และแถวที่ 1035 มาบวกกันทีละสมาชิกเป็นเวกเตอร์บัญชีครอบครัว s ทีมขายต้องการดูว่า "ทิศทาง" ของ s คล้ายกับลูกค้าต้นแบบ p = [66, 9, 7, 8] (ตัวแปรชุดเดียวกัน ลำดับเดียวกัน) เพียงใด ค่า cosine similarity ระหว่าง s กับ p =
0.9912
 ส่งดูเฉลยส่งแล้ว 1 ครั้ง✓ ถูกต้อง (ทศนิยม 4 ตำแหน่ง)**

```python
import pandas as pd
import numpy as np

# 1. โหลดไฟล์ข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์กำหนด:
# (รายได้ต่อเดือน, จำนวนอีเมลที่เปิดอ่าน, ขนาดตะกร้า, ยอดซื้อต่อเดือน)
cols = ['income_k', 'email_opens', 'basket_size', 'spend_k']

# 3. ดึงเวกเตอร์ลักษณะของลูกค้าแถวที่ 898 และแถวที่ 1035
vec_898 = df.iloc[898][cols].values.astype(np.float64)
vec_1035 = df.iloc[1035][cols].values.astype(np.float64)

# 4. รวมพฤติกรรมลูกค้าทั้งสองคนเข้าด้วยกัน (Element-wise addition) เกิดเป็นเวกเตอร์บัญชีครอบครัว s
s = vec_898 + vec_1035

# 5. เวกเตอร์ต้นแบบ p ที่โจทย์กำหนด
p = np.array([66, 9, 7, 8], dtype=np.float64)

# -------------------------------------------------------------
# 6. คำนวณ Cosine Similarity ระหว่าง s กับ p
# สูตร: Cosine Similarity = (s · p) / (||s|| * ||p||)
# -------------------------------------------------------------
dot_product = np.dot(s, p)             # ผลคูณจุด (Dot product)
norm_s = np.linalg.norm(s)             # ขนาด (L2 norm) ของ s
norm_p = np.linalg.norm(p)             # ขนาด (L2 norm) ของ p

cosine_sim = dot_product / (norm_s * norm_p)

# 7. แสดงผลลัพธ์
print(f"เวกเตอร์บัญชีครอบครัว s: {s}")
print(f"เวกเตอร์ต้นแบบ p: {p}")
print(f"Cosine Similarity (ทศนิยม 4 ตำแหน่ง): {cosine_sim:.4f}")
```

1.3
ให้ x = [age, income_k, email_opens, spend_k] เป็นเวกเตอร์ลักษณะ มุมระหว่างเวกเตอร์ของลูกค้าแถวที่ 1059 กับแถวที่ 509 มีขนาดกี่องศา
(ใช้ cosine similarity แล้วแปลงเป็นมุม) =
(ทศนิยม 2 ตำแหน่ง)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ: [age, income_k, email_opens, spend_k]
cols = ['age', 'income_k', 'email_opens', 'spend_k']

# 3. ดึงเวกเตอร์ลักษณะของลูกค้าแถวที่ 1059 และแถวที่ 509
vec_1059 = df.iloc[1059][cols].values.astype(np.float64)
vec_509 = df.iloc[509][cols].values.astype(np.float64)

# 4. คำนวณ Cosine Similarity
dot_product = np.dot(vec_1059, vec_509)
norm_1059 = np.linalg.norm(vec_1059)
norm_509 = np.linalg.norm(vec_509)

cos_sim = dot_product / (norm_1059 * norm_509)

# 5. แปลง Cosine Similarity เป็นมุมองศา
angle_rad = np.arccos(cos_sim)       # ได้มุมหน่วยเรเดียน
angle_deg = np.degrees(angle_rad)    # แปลงเป็นหน่วยองศา

# 6. แสดงผลลัพธ์
print(f"เวกเตอร์แถวที่ 1059: {vec_1059}")
print(f"เวกเตอร์แถวที่ 509:  {vec_509}")
print(f"Cosine Similarity:  {cos_sim:.6f}")
print(f"มุมองศา (ทศนิยม 2 ตำแหน่ง): {angle_deg:.2f}°")
```

ฝึก
ให้เวกเตอร์ลักษณะ 5 ตัวของลูกค้าแต่ละคนในไฟล์ biz_customers.csv คือ x = [income_k, tenure_months, web_visits, email_opens, orders_90d] (ตามลำดับนี้) ใช้ค่าตามไฟล์ (ไม่ standardize ; คำนวณด้วย float64 และไม่ปัดเศษ cosine ก่อนนำมาเปรียบเทียบ) หา cosine similarity ระหว่างลูกค้าแถวที่ 838 (นับแถวจาก 0 ตามลำดับในไฟล์) กับลูกค้าคนอื่นทุกคน จำนวนลูกค้า (ไม่นับแถวที่ 838 เอง) ที่มีค่า cosine similarity มากกว่า 0.9538 มี
ตัวเลข
 คน (จำนวนเต็ม)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ:
# [income_k, tenure_months, web_visits, email_opens, orders_90d]
cols = ['income_k', 'tenure_months', 'web_visits', 'email_opens', 'orders_90d']

# 3. แปลงเป็น NumPy Array แบบ float64
X = df[cols].values.astype(np.float64)

# 4. ดึงเวกเตอร์ของลูกค้าแถวที่ 838
x_838 = X[838]
norm_838 = np.linalg.norm(x_838)

# 5. คำนวณ Cosine Similarity กับลูกค้าทุกคนในไฟล์ (ใช้ Vectorization)
norms = np.linalg.norm(X, axis=1)        # ขนาด L2 norm ของลูกค้าทุกคน
dots = np.dot(X, x_838)                  # Dot product ระหว่างทุกคนกับแถวที่ 838
cos_sims = dots / (norms * norm_838)     # Cosine Similarity ของทุกคน

# 6. ตัดแถวที่ 838 เองออก (ไม่นับตัวเอง)
cos_sims_other = np.delete(cos_sims, 838)

# 7. นับจำนวนคนที่มี Cosine Similarity > 0.9538
count = np.sum(cos_sims_other > 0.9538)

print(f"เวกเตอร์แถวที่ 838: {x_838}")
print(f"จำนวนลูกค้าทั้งหมด (ไม่รวมตัวเอง): {len(cos_sims_other)}")
print(f"จำนวนลูกค้าที่มี Cosine Similarity > 0.9538: {count} คน")
```

โจทย์ข้อ 1.5 (กรณีศึกษา)
โจทย์: ธนาคารต้องการเสนอบัตรเครดิตใบใหม่ให้ลูกค้าที่ "ลักษณะคล้าย" ลูกค้าแถวที่ 118 ที่เพิ่งสมัครสำเร็จ โดยวัดความคล้ายจากระยะทางแบบยุคลิดระหว่างเวกเตอร์ลักษณะ (รายได้ต่อเดือน, อายุความเป็นลูกค้า, จำนวนอีเมลที่เปิดอ่าน) เรียงตามลำดับนี้ (ใช้หน่วยตามไฟล์ ไม่ปรับมาตราส่วน) ระยะทางที่สั้นที่สุดจากลูกค้าแถวที่ 118 ไปยังลูกค้าคนอื่นในไฟล์ (ไม่นับตัวเอง) = ส่งดูเฉลย (ทศนิยม 4 ตำแหน่ง)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ:
# [รายได้ต่อเดือน, อายุความเป็นลูกค้า, จำนวนอีเมลที่เปิดอ่าน]
cols = ['income_k', 'tenure_months', 'email_opens']

# 3. แปลงเป็น NumPy Array แบบ float64
X = df[cols].values.astype(np.float64)

# 4. ดึงเวกเตอร์ของลูกค้าแถวที่ 118
x_118 = X[118]

# 5. คำนวณระยะทางแบบยุคลิดจากแถวที่ 118 ไปยังทุกคนในไฟล์ (Broadcasting)
distances = np.linalg.norm(X - x_118, axis=1)

# 6. ตัดระยะทางของแถวที่ 118 ออก (ไม่นับตัวเอง ซึ่งระยะทางเป็น 0)
distances_other = np.delete(distances, 118)

# 7. หาระยะทางที่สั้นที่สุด (Minimum Distance)
min_distance = np.min(distances_other)

print(f"เวกเตอร์แถวที่ 118: {x_118}")
print(f"ระยะทางที่สั้นที่สุดไปยังลูกค้าคนอื่น: {min_distance:.4f}")
```

โจทย์ข้อ 1.6 (กรณีศึกษา)โจทย์: ทีมการตลาดดิจิทัลให้ "คะแนนความพร้อมซื้อ" แก่ลูกค้าแต่ละคนเป็นผลคูณจุด (dot product) ระหว่างเวกเตอร์ลักษณะ (รายได้ต่อเดือน, อายุความเป็นลูกค้า, ขนาดตะกร้า, จำนวนออเดอร์ใน 90 วัน) เรียงตามลำดับนี้ กับเวกเตอร์น้ำหนัก $\mathbf{w} = [0.2, 0.5, 0.1, 1.5]$ (ตามลำดับเดียวกัน) แล้วส่งโปรโมชันพิเศษให้ลูกค้าที่ได้คะแนนสูงสุด 3 อันดับแรกในไฟล์ คะแนนเฉลี่ยของลูกค้า 3 คนนี้ =  ส่งดูเฉลย (ทศนิยม 4 ตำแหน่ง)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ:
# [รายได้ต่อเดือน, อายุความเป็นลูกค้า, ขนาดตะกร้า, จำนวนออเดอร์ใน 90 วัน]
cols = ['income_k', 'tenure_months', 'basket_size', 'orders_90d']

# 3. แปลงเป็น NumPy Array แบบ float64
X = df[cols].values.astype(np.float64)

# 4. เวกเตอร์น้ำหนัก w
w = np.array([0.2, 0.5, 0.1, 1.5], dtype=np.float64)

# 5. คำนวณคะแนนความพร้อมซื้อของลูกค้าทุกคนด้วย Dot Product (Matrix-Vector Multiplication)
scores = np.dot(X, w)

# 6. คัดเลือกคะแนนสูงสุด 3 อันดับแรก
top3_scores = np.sort(scores)[-3:][::-1]

# 7. หาค่าเฉลี่ยของคะแนน 3 อันดับแรก
mean_top3 = np.mean(top3_scores)

print(f"คะแนนสูงสุด 3 อันดับแรก: {top3_scores}")
print(f"คะแนนเฉลี่ย (ทศนิยม 4 ตำแหน่ง): {mean_top3:.4f}")
```

โจทย์ข้อ 1.7 (กรณีศึกษา)
โจทย์: นักวิเคราะห์ของแพลตฟอร์มอีคอมเมิร์ซอยากรู้ว่าจำนวนครั้งที่เข้าเว็บ/แอปกับจำนวนออเดอร์ใน 90 วันของลูกค้า "ชี้ไปทิศทางเดียวกัน" มากแค่ไหน จึงนำค่าของลูกค้า 300 คนแรกในไฟล์ (แถวที่ 0 ถึง 299) มาทำเป็นเวกเตอร์ยาว 300 ตัวต่อหนึ่งตัวแปร (ไม่หักค่าเฉลี่ย) cosine similarity ระหว่างสองเวกเตอร์นี้ =  ส่งดูเฉลย (ทศนิยม 4 ตำแหน่ง)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. ดึงข้อมูล 300 แถวแรก (แถวที่ 0 ถึง 299) สำหรับแต่ละคอลัมน์
u = df.iloc[0:300]['web_visits'].values.astype(np.float64)  # เวกเตอร์ยาว 300
v = df.iloc[0:300]['orders_90d'].values.astype(np.float64)  # เวกเตอร์ยาว 300

# 3. คำนวณ Cosine Similarity ระหว่างเวกเตอร์ความยาว 300 มิติทั้งสอง
dot_product = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

cos_sim = dot_product / (norm_u * norm_v)

print(f"ความยาวเวกเตอร์ u: {len(u)}")
print(f"Dot Product (u · v): {dot_product:.1f}")
print(f"Norm of u (web_visits): {norm_u:.6f}")
print(f"Norm of v (orders_90d): {norm_v:.6f}")
print(f"Cosine Similarity (ทศนิยม 4 ตำแหน่ง): {cos_sim:.4f}")
```

โจทย์ข้อ 1.8 (ฝึก)โจทย์: เลือกลูกค้าแถวที่ 194, 425, 826, 839, 1064 (5 คน) แล้วสร้างเมทริกซ์ cosine similarity ขนาด $5 \times 5$ ระหว่างทุกคู่ จากเวกเตอร์ลักษณะ $\mathbf{x} = [\text{age}, \text{tenure\_months}, \text{web\_visits}, \text{spend\_k}]$ ค่า cosine similarity ที่ ต่ำที่สุด ในบรรดาคู่ต่างคน (ไม่นับแนวทแยง) =  ส่งดูเฉลย (ทศนิยม 4 ตำแหน่ง)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. เลือกลูกค้า 5 คนและคอลัมน์ตามที่โจทย์กำหนด
row_indices = [194, 425, 826, 839, 1064]
cols = ['age', 'tenure_months', 'web_visits', 'spend_k']

X_5 = df.iloc[row_indices][cols].values.astype(np.float64)

# 3. คำนวณ Cosine Similarity Matrix ขนาด 5x5
norms = np.linalg.norm(X_5, axis=1)
cos_matrix = np.dot(X_5, X_5.T) / np.outer(norms, norms)

# 4. หาค่าที่ต่ำที่สุดนอกแนวทแยงหลัก (Off-diagonal Minimum)
# เปลี่ยนแนวทแยงหลักให้เป็นค่าสูงๆ ชั่วคราวเพื่อไม่ให้ส่งผลต่อการหา min
matrix_no_diag = cos_matrix.copy()
np.fill_diagonal(matrix_no_diag, 999.0)

min_cos_sim = np.min(matrix_no_diag)

# 5. แสดงผลเมทริกซ์และค่าต่ำสุด
df_matrix = pd.DataFrame(cos_matrix, index=row_indices, columns=row_indices)
print("Cosine Similarity Matrix (5x5):")
print(df_matrix.round(4))

print(f"\nค่า Cosine Similarity ที่ต่ำที่สุดนอกแนวทแยง: {min_cos_sim:.4f}")
```

โจทย์ข้อ 1.9 (กรณีศึกษา)โจทย์: ร้านสะดวกซื้อกำหนดลูกค้าต้นแบบสำหรับโปรโมชันสมาชิกใหม่เป็นเวกเตอร์ $\mathbf{p} = [33, 8, 3]$ เรียงตาม (อายุความเป็นลูกค้า, จำนวนอีเมลที่เปิดอ่าน, ขนาดตะกร้า) ผู้จัดการจะส่งคูปองให้ลูกค้าที่เวกเตอร์ลักษณะ (ตัวแปรชุดเดียวกัน ลำดับเดียวกัน ใช้หน่วยตามไฟล์) อยู่ห่างจาก $\mathbf{p}$ ด้วยระยะทางแบบยุคลิดน้อยกว่า 9.2 มีลูกค้าในไฟล์ที่ได้รับคูปองกี่คน =  ส่งดูเฉลย คน (จำนวนเต็ม)

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ:
# [อายุความเป็นลูกค้า, จำนวนอีเมลที่เปิดอ่าน, ขนาดตะกร้า]
cols = ['tenure_months', 'email_opens', 'basket_size']

# 3. แปลงเป็น NumPy Array แบบ float64
X = df[cols].values.astype(np.float64)

# 4. เวกเตอร์ต้นแบบ p
p = np.array([33, 8, 3], dtype=np.float64)

# 5. คำนวณระยะทางแบบยุคลิดจากลูกค้าทุกคนไปยัง p (Broadcasting)
distances = np.linalg.norm(X - p, axis=1)

# 6. นับจำนวนลูกค้าที่มีระยะทางน้อยกว่า 9.2
count_coupon = np.sum(distances < 9.2)

print(f"เวกเตอร์ต้นแบบ p: {p}")
print(f"จำนวนลูกค้าทั้งหมดในไฟล์: {len(df)}")
print(f"จำนวนลูกค้าที่ได้รับคูปอง (ระยะทาง < 9.2): {count_coupon} คน")
```

โจทย์ข้อ 1.10 (ฝึก)โจทย์: ให้ $\mathbf{x} = [\text{age}, \text{income\_k}, \text{tenure\_months}, \text{orders\_90d}]$ เป็นเวกเตอร์ลักษณะของลูกค้า หาเวกเตอร์ค่าเฉลี่ยของลูกค้าที่ loyalty_member = 1 ทุกคนในไฟล์ และเวกเตอร์ค่าเฉลี่ยของลูกค้าที่ loyalty_member = 0 ทุกคน cosine similarity ระหว่างเวกเตอร์ค่าเฉลี่ยของสองกลุ่มนี้ =  ส่งดูเฉลย (ทศนิยม 4 ตำแหน่ง)1. แนวคิดและสูตรคณิตศาสตร์

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูล
df = pd.read_csv('biz_customers.csv')

# 2. กำหนดคอลัมน์ตามลำดับที่โจทย์ระบุ:
# [age, income_k, tenure_months, orders_90d]
cols = ['age', 'income_k', 'tenure_months', 'orders_90d']

# 3. แยกข้อมูลออกเป็น 2 กลุ่มตาม loyalty_member
group1 = df[df['loyalty_member'] == 1][cols].values.astype(np.float64)
group0 = df[df['loyalty_member'] == 0][cols].values.astype(np.float64)

# 4. หาเวกเตอร์ค่าเฉลี่ยของแต่ละกลุ่ม (Mean Vector / Centroid)
mean1 = np.mean(group1, axis=0)
mean0 = np.mean(group0, axis=0)

# 5. คำนวณ Cosine Similarity ระหว่างเวกเตอร์ค่าเฉลี่ยสองกลุ่ม
dot_product = np.dot(mean1, mean0)
norm1 = np.linalg.norm(mean1)
norm0 = np.linalg.norm(mean0)

cos_sim = dot_product / (norm1 * norm0)

print(f"เวกเตอร์ค่าเฉลี่ยกลุ่ม loyalty_member = 1: {mean1}")
print(f"เวกเตอร์ค่าเฉลี่ยกลุ่ม loyalty_member = 0: {mean0}")
print(f"Cosine Similarity (ทศนิยม 4 ตำแหน่ง): {cos_sim:.4f}")
```
