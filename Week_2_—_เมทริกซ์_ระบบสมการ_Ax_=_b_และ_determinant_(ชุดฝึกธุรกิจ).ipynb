โจทย์ข้อ 2.1โจทย์: ให้ $A$ คือเมทริกซ์ $6 \times 6$ ในไฟล์ pb02_A6.csv (แถวแรกเป็นชื่อคอลัมน์ c0–c5) โดยนิยาม $A^5 = A \cdot A \cdot A \cdot A \cdot A$ คือการคูณเมทริกซ์ $A$ กับตัวมันเองต่อกัน 5 ครั้ง ค่า $\text{trace}(A^5) =$ ? (ตอบเป็นจำนวนเต็ม)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ pb02_A6.csv ด้วย float64
A = pd.read_csv("pb02_A6.csv").to_numpy(dtype=np.float64)

# 2. คำนวณ A^5 ด้วยการคูณเมทริกซ์ (Matrix Power)
A5 = np.linalg.matrix_power(A, 5)  # หรือ A @ A @ A @ A @ A

# 3. คำนวณ Trace (ผลรวมเส้นทแยงมุมหลัก)
tr_A5 = np.trace(A5)

# 4. แปลงเป็นจำนวนเต็มตามที่โจทย์กำหนด
ans_2_1 = int(np.round(tr_A5))
print(f"ค่า trace(A^5) = {ans_2_1}")
```

โจทย์ข้อ 2.2โจทย์: ให้ $A$ (ขนาด $7 \times 7$) และ $b$ (ยาว $7$ ตัว) คือเมทริกซ์และเวกเตอร์ใน pb02_Asys.csv และ pb02_bsys.csv (แถวแรกของทั้งสองไฟล์เป็นชื่อคอลัมน์) แก้ระบบ $Ax = b$ แล้วสมาชิก $x[3]$ (ตัวที่ 3 ของเวกเตอร์ $x$ นับจาก 0) = ? (ตอบทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ pb02_Asys.csv และ pb02_bsys.csv ด้วยความละเอียด float64
A = pd.read_csv("pb02_Asys.csv").to_numpy(dtype=np.float64)
b = pd.read_csv("pb02_bsys.csv").to_numpy(dtype=np.float64).flatten()

# 2. แก้ระบบสมการ Ax = b
x = np.linalg.solve(A, b)

# 3. ดึงค่า x[3] และปัดเศษทศนิยม 4 ตำแหน่ง
ans_2_2 = round(x[3], 4)
print(f"ค่า x[3] = {ans_2_2:.4f}")
```

โจทย์ข้อ 2.3โจทย์: สร้างเมทริกซ์ $X$ ขนาด $25 \times 3$ จากแคมเปญ 25 แถวแรกในไฟล์ biz_campaigns.csv (แถวที่ 0 ถึง 24) โดยคอลัมน์ของ $X$ คือ radio, social, email (ตามลำดับนี้) แล้วคำนวณ $G = X^T X$ (ขนาด $3 \times 3$) ค่าสมาชิก $G[0, 2]$ (แถวที่ 0 คอลัมน์ที่ 2 นับจาก 0) = ? (ตอบทศนิยม 2 ตำแหน่ง)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ biz_campaigns.csv
df_campaigns = pd.read_csv("biz_campaigns.csv")

# 2. ดึง 25 แถวแรก (แถวที่ 0 ถึง 24) เฉพาะคอลัมน์ radio, social, email
X = df_campaigns.iloc[0:25][["radio", "social", "email"]].to_numpy(
    dtype=np.float64
)

# 3. คำนวณ G = X^T @ X
G = X.T @ X

# 4. ดึงค่า G[0, 2] และปัดเศษทศนิยม 2 ตำแหน่ง
ans_2_3 = round(G[0, 2], 2)
print(f"ค่า G[0, 2] = {ans_2_3:.2f}")
```

โจทย์ข้อ 2.4โจทย์: ทีมการตลาดอยากหาสมการ $\text{ยอดขาย} = \beta_0 + \beta_1 \cdot \text{งบทีวี} + \beta_2 \cdot \text{งบวิทยุ} + \beta_3 \cdot \text{งบโซเชียล}$ ที่ผ่านค่าของแคมเปญ 4 แคมเปญพอดี คือแถวที่ 63, 65, 196, 228 ในไฟล์แคมเปญ (แคมเปญหนึ่งให้หนึ่งสมการ รวม 4 สมการ 4 ตัวแปรที่ไม่ทราบค่า $\beta_0$–$\beta_3$) ค่า $\beta_3$ (สัมประสิทธิ์ของงบโฆษณาโซเชียล) = ? (ตอบทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ biz_campaigns.csv
df_campaigns = pd.read_csv("biz_campaigns.csv")

# 2. เลือกแถวที่ 63, 65, 196, 228
rows = [63, 65, 196, 228]
sub_df = df_campaigns.iloc[rows]

# 3. สร้างเมทริกซ์ X (ขนาด 4x4) และเวกเตอร์ y (ขนาด 4x1)
X = np.column_stack(
    [
        np.ones(4, dtype=np.float64),  # คอลัมน์สำหรับ Intercept (beta_0)
        sub_df[["tv", "radio", "social"]].to_numpy(dtype=np.float64),
    ]
)
y = sub_df["sales"].to_numpy(dtype=np.float64)

# 4. แก้ระบบสมการ X @ beta = y
beta = np.linalg.solve(X, y)

# 5. ดึงค่า beta_3 และปัดเศษทศนิยม 4 ตำแหน่ง
ans_2_4 = round(beta[3], 4)
print(f"ค่า beta_3 = {ans_2_4:.4f}")
```

โจทย์ข้อ 2.5โจทย์: ทำ Forward Elimination (ตามขั้นตอนใน Lab: ใช้แถวบนลบแถวล่างด้วย $\text{multiplier} = \text{สมาชิกใต้ pivot} \div \text{pivot}$, ไม่สลับแถว) กับเมทริกซ์ $E$ ขนาด $4 \times 4$ ต่อไปนี้ จนได้เมทริกซ์สามเหลี่ยมบน $U$ ค่า $U[3, 3]$ (สมาชิกแถวสุดท้าย คอลัมน์สุดท้าย) = ? (ตอบทศนิยม 4 ตำแหน่ง)

```python
import numpy as np

# 1. นิยามเมทริกซ์ E
E = np.array(
    [
        [2, 4, 6, -5],
        [-1, -3, -2, 6],
        [-1, -1, 1, 4],
        [-2, -1, -1, 5],
    ],
    dtype=np.float64,
)

U = E.copy()
n = U.shape[0]

# 2. ทำ Forward Elimination
for i in range(n):
    pivot = U[i, i]
    for j in range(i + 1, n):
        multiplier = U[j, i] / pivot
        U[j, i:] -= multiplier * U[i, i:]

# 3. ดึงค่า U[3, 3] และปัดเศษทศนิยม 4 ตำแหน่ง
ans_2_5 = round(U[3, 3], 4)
print(f"ค่า U[3, 3] = {ans_2_5:.4f}")
```

โจทย์ข้อ 2.6โจทย์: ทีมวิเคราะห์ต้องการทำนายยอดซื้อต่อเดือนจากอายุของลูกค้าด้วยสมการเส้นตรงที่มีค่าคงที่ จึงสร้างเมทริกซ์ออกแบบ $X$ ขนาด $25 \times 2$ จากลูกค้า 25 คนแรกในไฟล์ (แถวที่ 0 ถึง 24) คอลัมน์แรกเป็น 1 ทั้งหมด คอลัมน์ที่สองเป็นค่าอายุ แล้วตรวจว่า $X^T X$ (ขนาด $2 \times 2$) กลับด้านได้หรือไม่ ค่า $\det(X^T X) =$ ? (ตอบเป็นจำนวนเต็ม)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ biz_customers.csv
df_customers = pd.read_csv("biz_customers.csv")

# 2. ดึงข้อมูลอายุ (age) ของลูกค้า 25 คนแรก (แถวที่ 0 ถึง 24)
age = df_customers.iloc[0:25]["age"].to_numpy(dtype=np.float64)

# 3. สร้างเมทริกซ์ X ขนาด 25x2 (คอลัมน์แรกเป็น 1 ทั้งหมด, คอลัมน์สองคือ age)
X = np.column_stack([np.ones(25, dtype=np.float64), age])

# 4. คำนวณ X^T @ X
XTX = X.T @ X

# 5. คำนวณ determinant และแปลงเป็นจำนวนเต็ม
det_XTX = int(np.round(np.linalg.det(XTX)))
print(f"ค่า det(X^T X) = {det_XTX}")
```

โจทย์ข้อ 2.7โจทย์: ร้านค้าปลีกออนไลน์ใช้ข้อมูลลูกค้า 40 คนแรกในไฟล์ (แถวที่ 0 ถึง 39) ประมาณยอดซื้อต่อเดือนด้วยสมการ $\text{ยอดซื้อ} = \beta_0 + \beta_1 \cdot (\text{จำนวนครั้งที่เข้าเว็บ/แอป}) + \beta_2 \cdot (\text{จำนวนอีเมลที่เปิดอ่าน})$ โดยสร้างเมทริกซ์ $X$ ขนาด $40 \times 3$ (คอลัมน์: 1, จำนวนครั้งที่เข้าเว็บ/แอป, จำนวนอีเมลที่เปิดอ่าน ตามลำดับ) และเวกเตอร์ $y = \text{ยอดซื้อต่อเดือน}$ แล้วหา $\beta$ จากสมการปกติ $(X^T X)\beta = X^T y$ ค่า $\beta_2$ (สัมประสิทธิ์ของจำนวนอีเมลที่เปิดอ่าน) = ? (ตอบทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ biz_customers.csv เฉพาะ 40 คนแรก (แถวที่ 0 ถึง 39)
df_customers = pd.read_csv("biz_customers.csv").iloc[0:40]

# 2. สร้างเมทริกซ์ X (ขนาด 40x3) และเวกเตอร์ y (ขนาด 40x1)
X = np.column_stack(
    [
        np.ones(40, dtype=np.float64),
        df_customers["web_visits"].to_numpy(dtype=np.float64),
        df_customers["email_opens"].to_numpy(dtype=np.float64),
    ]
)
y = df_customers["spend_k"].to_numpy(dtype=np.float64)

# 3. แก้สมการปกติ (X^T @ X) @ beta = X^T @ y
XTX = X.T @ X
XTy = X.T @ y
beta = np.linalg.solve(XTX, XTy)

# 4. ดึงค่า beta_2 และปัดเศษทศนิยม 4 ตำแหน่ง
ans_2_7 = round(beta[2], 4)
print(f"ค่า beta_2 = {ans_2_7:.4f}")
```

โจทย์ข้อ 2.8โจทย์: ศูนย์กระจายสินค้าจัดสรรปริมาณสินค้า (พันชิ้น) ให้ร้านสาขา 7 แห่ง (สาขาที่ 0 ถึง 6) เป็นเวกเตอร์ $x$ ที่ต้องเป็นไปตามเงื่อนไข $Ax = b$ โดย $A$ ($7 \times 7$) และ $b$ คือเมทริกซ์และเวกเตอร์ใน pb02_Asys.csv และ pb02_bsys.csv ตอนนี้ผู้จัดการเพิ่มความต้องการในเงื่อนไขที่ 2 (สมาชิก $b[2]$) อีก 1 หน่วย โดย $b$ ตัวอื่นคงเดิม ปริมาณที่จัดสรรให้สาขาที่ 5 ($x[5]$) จะเปลี่ยนไปจากเดิมกี่พันชิ้น ($\text{ค่าใหม่} - \text{ค่าเดิม}$) = ? (ตอบทศนิยม 4 ตำแหน่ง ตอบเป็นค่าบวกหรือลบตามจริง)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ pb02_Asys.csv และ pb02_bsys.csv
A = pd.read_csv("pb02_Asys.csv").to_numpy(dtype=np.float64)
b = pd.read_csv("pb02_bsys.csv").to_numpy(dtype=np.float64).flatten()

# 2. หาคำตอบเดิม x_old
x_old = np.linalg.solve(A, b)

# 3. ปรับเพิ่มค่า b[2] อีก 1 หน่วย แล้วหา x_new
b_new = b.copy()
b_new[2] += 1.0
x_new = np.linalg.solve(A, b_new)

# 4. หาผลต่างของปริมาณที่จัดสรรให้สาขาที่ 5 (x[5])
delta_x5 = x_new[5] - x_old[5]

# 5. ปัดเศษทศนิยม 4 ตำแหน่ง
ans_2_8 = round(delta_x5, 4)
print(f"ปริมาณที่เปลี่ยนไปของสาขาที่ 5 (x[5]) = {ans_2_8:.4f}")
```

โจทย์ข้อ 2.9โจทย์: บริษัทสินเชื่อจัดบัญชีลูกหนี้เป็น 4 สถานะ และพบว่าในแต่ละเดือนบัญชีย้ายสถานะตามเมทริกซ์ $P$ ด้านล่าง (แถว = สถานะเดือนนี้, คอลัมน์ = สถานะเดือนถัดไป)เดือนนี้ → เดือนถัดไปชำระปกติล่าช้า 30 วันล่าช้า 60 วันค้างชำระเกิน 90 วันชำระปกติ0.000.400.200.40ล่าช้า 30 วัน0.450.500.050.00ล่าช้า 60 วัน0.200.450.300.05ค้างชำระเกิน 90 วัน0.600.000.100.30ให้ $p$ เป็นเวกเตอร์แถวของจำนวนบัญชีในแต่ละสถานะ เดือนนี้มี $p = [1200, 800, 1600, 300]$ บัญชี หลังผ่านไป 3 เดือน (คูณด้วยเมทริกซ์ $P$ จำนวน 3 ครั้ง) จะมีบัญชีในสถานะชำระปกติกี่บัญชี = ? (ตอบทศนิยม 2 ตำแหน่ง)

```python
import numpy as np

# 1. นิยามเมทริกซ์ P และเวกเตอร์สถานะเริ่มต้น p0
P = np.array(
    [
        [0.00, 0.40, 0.20, 0.40],
        [0.45, 0.50, 0.05, 0.00],
        [0.20, 0.45, 0.30, 0.05],
        [0.60, 0.00, 0.10, 0.30],
    ],
    dtype=np.float64,
)

p0 = np.array([1200, 800, 1600, 300], dtype=np.float64)

# 2. คูณเมทริกซ์ 3 ครั้งเพื่อหา p3
p1 = p0 @ P
p2 = p1 @ P
p3 = p2 @ P

# 3. ดึงค่าสถานะชำระปกติ p3[0] และปัดเศษทศนิยม 2 ตำแหน่ง
ans_2_9 = round(p3[0], 2)
print(f"จำนวนบัญชีชำระปกติหลังผ่านไป 3 เดือน = {ans_2_9:.2f}")
```

โจทย์ข้อ 2.10โจทย์: ให้ $A$ คือเมทริกซ์ $6 \times 6$ ใน pb02_A6.csv และ $S = A + A^T$ ($A^T$ คือทรานสโพสของ $A$) หาเมทริกซ์ $S \cdot S$ (คูณเมทริกซ์ด้วย @ ไม่ใช่ยกกำลังทีละสมาชิก) ค่า $\text{trace}(S \cdot S) =$ ? (ตอบเป็นจำนวนเต็ม)

```python
import numpy as np
import pandas as pd

# 1. อ่านไฟล์ pb02_A6.csv ด้วยความละเอียด float64
A = pd.read_csv("pb02_A6.csv").to_numpy(dtype=np.float64)

# 2. สร้างเมทริกซ์ S = A + A^T
S = A + A.T

# 3. คำนวณ S @ S
SS = S @ S

# 4. หาค่า Trace และแปลงเป็นจำนวนเต็ม
tr_SS = np.trace(SS)
ans_2_10 = int(np.round(tr_SS))

print(f"ค่า trace(S*S) = {ans_2_10}")
```

# สรุปทฤษฎีบทและแนวคิดคณิตศาสตร์ (Week 2: เมทริกซ์, ระบบสมการ $Ax = b$ และ Determinant)

---

## 1. ทฤษฎีบทระบบสมการเชิงเส้น ($Ax = b$)

* **ทฤษฎีบทการมีคำตอบเดียว (Uniqueness Theorem):**
  สำหรับเมทริกซ์จัตุรัส $A$ ขนาด $n \times n$ ระบบสมการ $Ax = b$ จะมีคำตอบเดียว $x$ ก็ต่อเมื่อ $A$ เป็นเมทริกซ์ไม่อ่อนพฤติ (Nonsingular Matrix) หรือ $\det(A) \neq 0$
  $$x = A^{-1} b$$

* **สมบัติการเปลี่ยนแปลงเชิงเส้น (Linear Perturbation Property):**
  เมื่อเวกเตอร์ด้านขวา $b$ เปลี่ยนไปเป็น $b + \Delta b$ จะส่งผลให้เวกเตอร์คำตอบเปลี่ยนไป $\Delta x$ ตามสมการ:
  $$A(x + \Delta x) = b + \Delta b \implies A \Delta x = \Delta b \implies \Delta x = A^{-1} \Delta b$$

---

## 2. การกำจัดไปข้างหน้า (Forward Elimination)

* **แนวคิด:** แปลงระบบสมการ $Ax = b$ ให้อยู่ในรูปเมทริกซ์สามเหลี่ยมบน $Ux = c$ โดยกำจัดสมาชิกใต้เส้นทแยงมุมหลักให้เป็น $0$ ทีละคอลัมน์จากซ้ายไปขวา
* **สูตรตัวคูณ (Multiplier):** สำหรับการกำจัดสมาชิกในแถวที่ $j$ คอลัมน์ที่ $i$ โดยใช้ $U_{i,i}$ เป็นจุดหมุน (Pivot):
  $$m_{j, i} = \frac{U_{j, i}}{U_{i, i}}$$
  $$R_j \leftarrow R_j - m_{j, i} R_i$$

## 3. ดีเทอร์มิแนนต์ (Determinant) และการกลับด้านได้

* **เงื่อนไขการหาอินเวิร์ส:** เมทริกซ์ $M$ จะหาอินเวิร์ส ($M^{-1}$) ได้ ก็ต่อเมื่อ $\det(M) \neq 0$
* **ดีเทอร์มิแนนต์ของเมทริกซ์ $2 \times 2$:**
  $$\det \begin{bmatrix} a & b \\ c & d \end{bmatrix} = ad - bc$$
* **สมบัติสำคัญ:**
  1. $\det(A^T) = \det(A)$
  2. $\det(AB) = \det(A) \cdot \det(B)$
  3. $\det(A^{-1}) = \frac{1}{\det(A)}$

## 4. สมการปกติและการประมาณค่าด้วยวิธีกำลังสองน้อยที่สุด (Normal Equations)

* **สมการปกติ (Normal Equations):**
  เมื่อระบบสมการ $X \beta = y$ มีจำนวนสมการมากกว่าตัวแปร ($m > n$) และไม่มีคำตอบที่สอดคล้องพอดีทุกจุด การหา $\beta$ ที่ให้ค่าความคลาดเคลื่อนกำลังสองน้อยที่สุด (OLS) คำนวณได้จาก:
  $$(X^T X)\beta = X^T y \implies \beta = (X^T X)^{-1} X^T y$$
* **เมทริกซ์ Gram ($G = X^T X$):**
  เป็นเมทริกซ์สมมาตร ($G = G^T$) ขนาด $n \times n$ โดยมีสมาชิก:
  $$G_{i, j} = X_{*, i}^T X_{*, j} = \sum_{k=0}^{m-1} X_{k, i} X_{k, j}$$

## 5. กระบวนการมาร์คอฟและการย้ายสถานะ (Markov Chain)

* **ความสัมพันธ์เชิงเวลา:**
  กำหนดให้ $p^{(0)}$ เป็นเวกเตอร์แถวของสถานะเริ่มต้น และ $P$ เป็น Transition Matrix (ผลรวมสมาชิกในแต่ละแถวเท่ากับ 1)
* **การคำนวณสถานะ ณ เวลา $k$:**
  $$p^{(k)} = p^{(k-1)} P = p^{(0)} P^k$$

## 6. สมบัติของเทรซ (Trace Properties)

* **นิยาม:** ผลรวมของสมาชิกบนเส้นทแยงมุมหลักของเมทริกซ์จัตุรัส $A$ ขนาด $n \times n$:
  $$\text{trace}(A) = \sum_{i=0}^{n-1} A_{i,i}$$
* **สมบัติพีชคณิต:**
  1. $\text{trace}(A + B) = \text{trace}(A) + \text{trace}(B)$
  2. $\text{trace}(c A) = c \cdot \text{trace}(A)$
  3. $\text{trace}(A^T) = \text{trace}(A)$
  4. $\text{trace}(AB) = \text{trace}(BA)$
