**Week 8 — Simple Linear Regression (SE, RSE, t, p-value, พยากรณ์, Test MSE)**

### สรุปเงื่อนไขการประมวลผลข้อมูล
- **การอ่านไฟล์:**
  - ข้อมูลลูกค้า: `cust = pd.read_csv('biz_customers.csv')` (1203 แถว)
  - ข้อมูลแคมเปญ: `camp = pd.read_csv('biz_campaigns.csv')` (240 แถว)
- **การนับดัชนี (0-based Indexing):** เริ่มนับจาก 0 ทั้งแถวและคอลัมน์
- **การเลือกช่วงแถว:** ข้อที่ระบุ "แถวที่ a ถึง b" ให้ใช้ `.iloc[a:b+1]` (รวมแถว a และ b)
- **แบบจำลอง Simple Linear Regression (SLR):** มี intercept $\hat{\beta}_0$ เสมอ และประมาณค่าด้วย Ordinary Least Squares (OLS)
- **สูตรทางสถิติ:**
  - $\text{RSS} = \sum_{i=1}^n (y_i - \hat{y}_i)^2$
  - $\text{RSE} = \sqrt{\frac{\text{RSS}}{n - 2}}$
  - $\text{SE}(\hat{\beta}_1) = \frac{\text{RSE}}{\sqrt{\sum (x_i - \bar{x})^2}}$
  - $t = \frac{\hat{\beta}_1}{\text{SE}(\hat{\beta}_1)}$ โดยมี $df = n - 2$

### รายละเอียดไฟล์ข้อมูล

**1. `biz_customers.csv`** (1203 แถว แถวละ 1 คน)
- `gender`: เพศของลูกค้า (`F` = หญิง, `M` = ชาย)
- `age`: อายุของลูกค้า (ปี)
- `income_k`: รายได้ต่อเดือน (พันบาท)
- `tenure_months`: ระยะเวลาที่เป็นลูกค้า (เดือน)
- `channel`: ช่องทางที่ซื้อบ่อย (`online` = ออนไลน์, `store` = หน้าร้าน)
- `web_visits`: จำนวนครั้งที่เข้าเว็บ/แอปใน 30 วันล่าสุด
- `spend_k`: ยอดซื้อต่อเดือน (พันบาท)

**2. `biz_campaigns.csv`** (240 แถว แถวละ 1 แคมเปญ)
- `tv`: งบโฆษณาทางโทรทัศน์ (พันบาท)
- `radio`: งบโฆษณาทางวิทยุ (พันบาท)
- `social`: งบโฆษณาทางโซเชียลมีเดีย (พันบาท)
- `email`: งบโฆษณาทางอีเมล (พันบาท)
- `total_budget`: งบโฆษณารวมทุกสื่อ (พันบาท)
- `season`: ไตรมาสของแคมเปญ (`Q1`–`Q4`)
- `sales`: ยอดขายของแคมเปญ (พันชิ้น)

**โจทย์ข้อ 8.1:** ฝึกใช้ไฟล์ `biz_customers.csv` ทำ SLR ที่ให้ `spend_k` เป็น $y$ และ `web_visits` เป็น $x$ (มี intercept) ของลูกค้าทั้ง 1203 คนในไฟล์ ประมาณด้วย Least Squares ค่า Standard Error ของความชัน $\text{SE}(\hat{\beta}_1)$ = ? (ทศนิยม 5 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. อ่านข้อมูลลูกค้าทั้งหมด 1203 คน
cust = pd.read_csv('biz_customers.csv')
x = cust['web_visits']
y = cust['spend_k']

# 2. ทำ SLR ด้วย scipy.stats.linregress
res = stats.linregress(x, y)
print(f"ความชัน (slope): {res.slope:.6f}")
print(f"Standard Error ของความชัน SE(beta1): {res.stderr:.5f}")
print(f"คำตอบข้อ 8.1: {res.stderr:.5f}")
```

**โจทย์ข้อ 8.2:** ฝึกใช้ไฟล์ `biz_campaigns.csv` ทำ SLR ที่ให้ `sales` เป็น $y$ และ `social` เป็น $x$ (มี intercept) ของแคมเปญทั้ง 240 แคมเปญในไฟล์ ประมาณด้วย Least Squares ค่า Residual Standard Error (RSE) = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. อ่านข้อมูลแคมเปญทั้ง 240 แถว
camp = pd.read_csv('biz_campaigns.csv')
x = camp['social']
y = camp['sales']
n = len(camp)

# 2. ประมาณค่า SLR
res = stats.linregress(x, y)
y_hat = res.intercept + res.slope * x

# 3. คำนวณ Residual Sum of Squares (RSS) และ RSE = sqrt(RSS / (n - 2))
rss = np.sum((y - y_hat)**2)
rse = np.sqrt(rss / (n - 2))
print(f"RSS: {rss:.4f}")
print(f"RSE: {rse:.4f}")
print(f"คำตอบข้อ 8.2: {rse:.4f}")
```

**โจทย์ข้อ 8.3:** ฝึกใช้เฉพาะลูกค้าที่ซื้อบ่อยที่หน้าร้าน (`channel == 'store'`) ในตาราง cust ทำ SLR ที่ให้ `spend_k` เป็น $y$ และ `tenure_months` เป็น $x$ ค่าประมาณ intercept $\hat{\beta}_0$ = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. กรองเฉพาะลูกค้าช่องทางหน้าร้าน (channel == 'store')
cust = pd.read_csv('biz_customers.csv')
cust_store = cust[cust['channel'] == 'store']
print(f"จำนวนลูกค้าหน้าร้าน: {len(cust_store)} คน")

# 2. ทำ SLR ที่ให้ y = spend_k, x = tenure_months
res = stats.linregress(cust_store['tenure_months'], cust_store['spend_k'])
print(f"Intercept (beta0): {res.intercept:.4f}")
print(f"Slope (beta1): {res.slope:.4f}")
print(f"คำตอบข้อ 8.3: {res.intercept:.4f}")
```

**โจทย์ข้อ 8.4:** กรณีศึกษาผู้จัดการแบรนด์เครื่องดื่มวางแผนแคมเปญของไตรมาส Q1 จึงสร้าง SLR ที่ให้ยอดขาย (พันชิ้น) `sales` เป็น $y$ และงบโฆษณาโทรทัศน์ (พันบาท) `tv` เป็น $x$ (มี intercept) จากแคมเปญที่ทำในไตรมาส Q1 (`season == 'Q1'`) ทุกแคมเปญในตาราง camp แล้วประมาณยอดขายของแคมเปญใหม่ที่ตั้งงบ $\text{tv} = 185$ พันบาท ค่า sales ที่โมเดลพยากรณ์ = ? (ทศนิยม 3 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. กรองแคมเปญเฉพาะไตรมาส Q1
camp = pd.read_csv('biz_campaigns.csv')
camp_q1 = camp[camp['season'] == 'Q1']
print(f"จำนวนแคมเปญใน Q1: {len(camp_q1)}")

# 2. สร้างโมเดล SLR: y = sales, x = tv
res = stats.linregress(camp_q1['tv'], camp_q1['sales'])
print(f"สมการพยากรณ์: y = {res.intercept:.4f} + {res.slope:.4f} * tv")

# 3. พยากรณ์ที่งบ tv = 185
tv_new = 185.0
pred_sales = res.intercept + res.slope * tv_new
print(f"ยอดขายที่พยากรณ์ได้: {pred_sales:.3f}")
print(f"คำตอบข้อ 8.4: {pred_sales:.3f}")
```

**โจทย์ข้อ 8.5:** ฝึกใช้ตาราง camp ทุกแถว (240 แถว) ทำ SLR ที่ให้ `sales` เป็น $y$ และ `total_budget` เป็น $x$ (มี intercept) ค่า t-statistic ของความชัน (ทดสอบ $H_0: \beta_1 = 0$) = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. อ่านข้อมูลแคมเปญทั้งหมด
camp = pd.read_csv('biz_campaigns.csv')
x = camp['total_budget']
y = camp['sales']

# 2. ทำ SLR และคำนวณ t-statistic ของความชัน
res = stats.linregress(x, y)
t_stat = res.slope / res.stderr
print(f"Slope: {res.slope:.6f}, SE(Slope): {res.stderr:.6f}")
print(f"t-statistic: {t_stat:.4f}")
print(f"คำตอบข้อ 8.5: {t_stat:.4f}")
```

**โจทย์ข้อ 8.6:** กรณีศึกษาทีมวิเคราะห์ของเอเจนซี่โฆษณามีข้อมูลนำร่องเพียง 30 แถว (**แถวที่ 157 ถึง 186** ของตาราง camp) จึงทำ SLR ที่ให้ยอดขาย (พันชิ้น) `sales` เป็น $y$ และงบโฆษณาโซเชียลมีเดีย (พันบาท) `social` เป็น $x$ (มี intercept) เพื่อทดสอบ $H_0: \beta_1 = 0$ แบบสองด้านด้วย t-distribution ($df = n - 2$) ค่า p-value ของความชัน = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. เลือกแถวที่ 157 ถึง 186 (.iloc[157:187], 30 แถว)
camp = pd.read_csv('biz_campaigns.csv')
sub_camp = camp.iloc[157:187]
print(f"ขนาดชุดข้อมูลนำร่อง: {sub_camp.shape}")

# 2. ทำ SLR ทดสอบ H0: beta1 = 0 (two-sided)
res = stats.linregress(sub_camp['social'], sub_camp['sales'])
print(f"t-statistic: {res.slope / res.stderr:.4f}")
print(f"p-value ของความชัน: {res.pvalue:.4f}")
print(f"คำตอบข้อ 8.6: {res.pvalue:.4f}")
```

**โจทย์ข้อ 8.7:** กรณีศึกษานักวิเคราะห์ของห้างสรรพสินค้าสร้าง SLR ที่ให้ยอดซื้อต่อเดือน (พันบาท) `spend_k` เป็น $y$ และจำนวนครั้งที่เข้าเว็บ/แอปใน 30 วัน `web_visits` เป็น $x$ (มี intercept) แยกตามเพศ เพื่อเปรียบเทียบความคลาดเคลื่อนของโมเดล สำหรับโมเดลของ**ลูกค้าเพศหญิง (`gender == 'F'`) ทุกคน 602 คน**ในตาราง cust ผลรวมกำลังสองของส่วนเหลือ **RSS** = ? (ทศนิยม 2 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. กรองเฉพาะลูกค้าเพศหญิง (gender == 'F')
cust = pd.read_csv('biz_customers.csv')
cust_f = cust[cust['gender'] == 'F']
print(f"จำนวนลูกค้าเพศหญิง: {len(cust_f)} คน")

# 2. ทำ SLR: y = spend_k, x = web_visits
res = stats.linregress(cust_f['web_visits'], cust_f['spend_k'])
y_hat = res.intercept + res.slope * cust_f['web_visits']

# 3. คำนวณ Residual Sum of Squares (RSS)
residuals = cust_f['spend_k'] - y_hat
rss = np.sum(residuals**2)
print(f"ผลรวมกำลังสองของส่วนเหลือ RSS: {rss:.2f}")
print(f"คำตอบข้อ 8.7: {rss:.2f}")
```

**โจทย์ข้อ 8.8:** ฝึกใช้ตาราง camp ทุกแถว (240 แถว) ทำ SLR ที่ให้ `sales` เป็น $y$ และ `email` เป็น $x$ (มี intercept) ส่วนเหลือ (residual) $e_i = y_i - \hat{y}_i$ ของ**แถวที่ 233** (นับแถวแรกเป็นแถวที่ 0) = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. อ่านข้อมูลแคมเปญทั้งหมด 240 แถว
camp = pd.read_csv('biz_campaigns.csv')
x = camp['email']
y = camp['sales']

# 2. ทำ SLR: y = sales, x = email
res = stats.linregress(x, y)
y_hat = res.intercept + res.slope * x

# 3. คำนวณส่วนเหลือ e_i = y_i - y_hat_i ที่แถว 233
residuals = y - y_hat
e_233 = residuals.iloc[233]
print(f"แถวที่ 233: y_จริง = {y.iloc[233]:.2f}, y_พยากรณ์ = {y_hat.iloc[233]:.4f}")
print(f"ส่วนเหลือ e_233: {e_233:.4f}")
print(f"คำตอบข้อ 8.8: {e_233:.4f}")
```

**โจทย์ข้อ 8.9:** กรณีศึกษาทีมข้อมูลของบริษัทค้าปลีกใช้ลูกค้า 960 แถวแรกของตาราง cust (แถวที่ 0 ถึง 959) เป็นชุดสอนโมเดล SLR ที่ให้ยอดซื้อต่อเดือน (พันบาท) `spend_k` เป็น $y$ และรายได้ต่อเดือน (พันบาท) `income_k` เป็น $x$ (มี intercept) แล้วนำโมเดลไปพยากรณ์ลูกค้าที่เหลือ (แถวที่ 960 ถึง 1202, 243 แถว) ค่า **Test MSE** = ? (ทศนิยม 4 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 1. แบ่งข้อมูล train (960 แถวแรก) และ test (243 แถวที่เหลือ)
cust = pd.read_csv('biz_customers.csv')
train = cust.iloc[:960]
test = cust.iloc[960:]
print(f"ขนาดชุด Train: {train.shape}, ขนาดชุด Test: {test.shape}")

# 2. Fit โมเดล SLR ด้วยชุด train
model = LinearRegression().fit(train[['income_k']], train['spend_k'])

# 3. พยากรณ์บนชุด test และคำนวณ Test MSE
pred_test = model.predict(test[['income_k']])
test_mse = mean_squared_error(test['spend_k'], pred_test)
print(f"Test MSE: {test_mse:.4f}")
print(f"คำตอบข้อ 8.9: {test_mse:.4f}")
```

**โจทย์ข้อ 8.10:** กรณีศึกษายอดซื้อต่อเดือนของลูกค้ามีการกระจายเบ้ขวา ทีมวิเคราะห์ของธุรกิจสมัครสมาชิกจึงสร้าง SLR ที่ให้ $y = \ln(\text{spend\_k})$ (ลอการิทึมธรรมชาติของยอดซื้อต่อเดือน พันบาท ใช้ `np.log`) และ $x$ = รายได้ต่อเดือน (พันบาท) `income_k` (มี intercept) จากลูกค้าทุกคน 1203 คนในตาราง cust แล้วพยากรณ์ลูกค้ารายใหม่ที่มี $\text{income\_k} = 25$ ค่าพยากรณ์ยอดซื้อต่อเดือนในหน่วยพันบาท (แปลงกลับด้วย `np.exp` จากค่าพยากรณ์ของ $\ln(\text{spend\_k})$) = ? (ทศนิยม 3 ตำแหน่ง)

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. อ่านข้อมูลลูกค้าและแปลง y ด้วย natural logarithm (np.log)
cust = pd.read_csv('biz_customers.csv')
x = cust['income_k']
y_log = np.log(cust['spend_k'])

# 2. Fit SLR บน log-transformed data
res = stats.linregress(x, y_log)
print(f"สมการในรูป log: ln(spend_k) = {res.intercept:.4f} + {res.slope:.4f} * income_k")

# 3. พยากรณ์ที่ income_k = 25 แล้วแปลงกลับด้วย np.exp
pred_log = res.intercept + res.slope * 25.0
pred_orig = np.exp(pred_log)
print(f"ค่าพยากรณ์ ln(spend_k): {pred_log:.4f}")
print(f"ค่าพยากรณ์ spend_k (หน่วยพันบาท): {pred_orig:.3f}")
print(f"คำตอบข้อ 8.10: {pred_orig:.3f}")
```

### 1. แบบจำลอง Simple Linear Regression (SLR)

แบบจำลองการถดถอยเชิงเส้นอย่างง่ายแสดงความสัมพันธ์ระหว่างตัวแปรทำนายเชิงปริมาณ $X$ และตัวแปรตอบสนอง $Y$:
$$Y = \beta_0 + \beta_1 X + \epsilon, \quad \mathbb{E}[\epsilon] = 0, \quad \text{Var}(\epsilon) = \sigma^2$$

**การประมาณค่าพารามิเตอร์ด้วย Ordinary Least Squares (OLS):**
$$\hat{\beta}_1 = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2} = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}$$
$$\hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x}$$

---

### 2. ผลรวมกำลังสองของส่วนเหลือ (RSS) และ Residual Standard Error (RSE)

- **ส่วนเหลือ (Residual):** $e_i = y_i - \hat{y}_i$
- **Residual Sum of Squares (RSS):**
  $$\text{RSS} = \sum_{i=1}^n e_i^2 = \sum_{i=1}^n (y_i - \hat{y}_i)^2$$
- **Residual Standard Error (RSE):** คือการประมาณค่าส่วนเบี่ยงเบนมาตรฐานของ error term ($\sigma$):
  $$\text{RSE} = \sqrt{\frac{\text{RSS}}{n - 2}}$$
  โดยหารด้วย $n - 2$ เนื่องจากสูญเสียองศาความเป็นอิสระไป 2 หน่วยจากการประมาณ $\beta_0$ และ $\beta_1$

---

### 3. Standard Errors และการทดสอบสมมติฐานบนความชัน

- **Standard Error ของสัมประสิทธิ์ความชัน:**
  $$\text{SE}(\hat{\beta}_1) = \frac{\text{RSE}}{\sqrt{\sum_{i=1}^n (x_i - \bar{x})^2}}$$
- **การทดสอบสมมติฐาน $H_0: \beta_1 = 0$ เทียบกับ $H_1: \beta_1 \ne 0$:**
  $$t = \frac{\hat{\beta}_1 - 0}{\text{SE}(\hat{\beta}_1)}$$
  ภายใต้สมมติฐานหลัก สถิติ $t$ มีการแจกแจงแบบ $t$-distribution ด้วยองศาความเป็นอิสระ $df = n - 2$
  - ค่า $|t|$ สูง และ $p\text{-value} < 0.05$ แสดงว่ามีความสัมพันธ์เชิงเส้นระหว่าง $X$ และ $Y$ อย่างมีนัยสำคัญ

---

### 4. การแปลงรูปตัวแปร (Log-linear Transformation)

เมื่อข้อมูลตัวแปรตาม $Y$ มีการแจกแจงที่เบ้ขวาหรือมีความแปรปรวนไม่คงที่ (Heteroscedasticity) นิยมใช้การแปลงรูปด้วยลอการิทึมธรรมชาติ:
$$\ln(Y) = \beta_0 + \beta_1 X + \epsilon$$
- การแปลงกลับเพื่อพยากรณ์ค่า $Y$ ในหน่วยดั้งเดิม:
  $$\hat{Y} = \exp\left(\widehat{\ln(Y)}\right) = \exp(\hat{\beta}_0 + \hat{\beta}_1 X)$$
